```python
import sys
sys.path.append("../src")

import numpy as np
import torch
from sklearn.utils.class_weight import compute_class_weight
from veri import yukleyici
from modeller import basit_cnn
from egitim import egit
from degerlendirme import olc, kaydet, grafik_ciz

csv = "../splits/veri2.csv"
tr = yukleyici(csv, "egitim", egit=True)
dg = yukleyici(csv, "dogrulama")
te = yukleyici(csv, "test")
siniflar = tr.dataset.siniflar
w = compute_class_weight("balanced", classes=np.array(siniflar), y=tr.dataset.df["sinif"])
w = torch.tensor(w, dtype=torch.float32)

model, gecmis = egit(basit_cnn(len(siniflar)), tr, dg, w)
sonuc, kar = olc(model, te, siniflar)
kaydet("v2_cnn", model, gecmis, sonuc, kar)
print(sonuc)
print(kar)
grafik_ciz("v2_cnn")
```
